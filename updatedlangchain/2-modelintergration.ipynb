### GOOGLE GENAI

```python
import os
from langchain.chat_models import init_chat_model

os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")

model = init_chat_model("google_genai:gemini-2.5-flash-lite")
response = model.invoke("Why do parrot talks?")
response
```

### GROQ

```python
import os
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = init_chat_model("groq:qwen/qwen3-32b")
response = model.invoke("Why do parrot talks?")
response
```

```python
import os
from langchain_groq import ChatGroq

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = ChatGroq(model="qwen/qwen3-32b")
response = model.invoke("Why do parrot talks?")
response
```

### Streaming

```python
model.stream("Write me a paragraph on neural networks")
```

```python
for chunck in model.stream("Write me a paragraph on neural networks"):
    print(chunck.text, end='|', flush=True)
```

### Batch

```python
responses = model.batch([
    "Why are parrots are talking?",
    "What is a llm?",
    "What is a neural network?"
])

for response in responses:
    print(response)
```

```python
responses = model.batch([
    "Why are parrots are talking?",
    "What is a llm?",
    "What is a neural network?"],
    config={
        'max_currency':5 #limit to 5 parralel calls
    }

)

for response in responses:
    print(response)
```
